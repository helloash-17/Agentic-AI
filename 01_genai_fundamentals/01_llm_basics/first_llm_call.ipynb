## 📚 Concept 1 — What is a Language Model?

A language model is an AI system trained to understand and generate human-like text.

At a basic level, it predicts the next word or token based on the context it receives.

### Basic idea

User Input
    ↓
Language Model
    ↓
Generated Response

The model uses the input context to determine what response to generate.

### Example

Input:

"What is an AI agent?"

Output:

The model generates a response based on the context and patterns it learned during training.
