## 🔌 Create the Groq Client

# 🚀 Experiment 1 — Our First LLM Call

Now we will send our first prompt to a language model through the Groq API.

### Design

Python Notebook
      │
      │ Prompt
      ↓
   Groq API
      │
      ↓
 Language Model
      │
      ↓
   Response
      │
      ↓
Python Notebook

### Our experiment

We will ask:

> What is an AI agent?

Then we will inspect the response returned by the model.

```python
response = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[
        {
            "role": "user",
            "content": "What is an AI agent?"
        }
    ]
)

print(response.choices[0].message.content)
```

## 🔍 What Just Happened?

Our Python program:

1. Created a request.
2. Sent the request to the Groq API.
3. The Groq API passed the request to the selected language model.
4. The model processed the input.
5. The model generated a response.
6. The response was returned to Python.
7. We printed the generated text.

### Important distinction

Groq is the **API/inference platform** we are using.

The language model is the **model generating the response**.

In this experiment, we are using:

**Llama 3.1 8B Instant**

### Basic flow

Python
  ↓
Groq API
  ↓
Llama 3.1 8B Instant
  ↓
Response

# 🧪 Experiment 2 — Changing the Prompt

We will keep the model the same and change only the prompt.

### Model

`llama-3.1-8b-instant`

### Prompt A

"What is an AI agent?"

### Prompt B

"Explain an AI agent in one sentence."

### Prompt C

"Explain an AI agent to a 10-year-old."

### Prompt D

"Explain an AI agent to a Python developer."

### Question

What changes?

- Model? ❌
- API provider? ❌
- Prompt? ✅
- Response? ✅

Let's run the experiment and observe the difference.

```python
prompts = [
    "What is an AI agent?",
    "Explain an AI agent in one sentence.",
    "Explain an AI agent to a 10-year-old.",
    "Explain an AI agent to a Python developer."
]

for prompt in prompts:

    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    print("\n" + "=" * 80)
    print("PROMPT:")
    print(prompt)

    print("\nRESPONSE:")
    print(response.choices[0].message.content)
```
